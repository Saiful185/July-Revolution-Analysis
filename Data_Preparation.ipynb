# **A Mixed-Methods Analysis of Repression and Mobilization in Bangladesh's July Revolution**

This notebook contains the full Data Preparation codes from the ACLED dataset during the July Revolution. These modified datasets will be used for ML and statistical modeeling based analysis.

## **Importing Libraries**

```python
import pandas as pd
import numpy as np
```

# **Modiefied Daily Time Series Dataset Preparation from Raw Dataset**

```python
# Loading Raw dataset
file_path = '/content/drive/MyDrive/Research/Bangladesh Movement Research/2024-06-05-2024-08-05-South_Asia-Bangladesh.csv'
data = pd.read_csv(file_path)

# Converting event_date to datetime
data['event_date'] = pd.to_datetime(data['event_date'])

# Grouping by event_date and aggregating the counts
grouped_data = data.groupby('event_date').agg(
    event_type_counts=pd.NamedAgg(column='event_type', aggfunc=lambda x: x.value_counts().to_dict()),
    sub_event_type_counts=pd.NamedAgg(column='sub_event_type', aggfunc=lambda x: x.value_counts().to_dict()),
    disorder_type_counts=pd.NamedAgg(column='disorder_type', aggfunc=lambda x: x.value_counts().to_dict()),
    actor1_counts=pd.NamedAgg(column='actor1', aggfunc=lambda x: x.value_counts().to_dict()),
    total_fatalities=pd.NamedAgg(column='fatalities', aggfunc='sum'),
    total_events=pd.NamedAgg(column='event_id_cnty', aggfunc='count')
).reset_index()

# Expanding the dictionaries into separate columns
expanded_data = grouped_data.apply(lambda row: pd.Series({
    **row['event_type_counts'],
    **row['sub_event_type_counts'],
    **row['disorder_type_counts'],
    **row['actor1_counts'],
    'total_fatalities': row['total_fatalities'],
    'total_events': row['total_events']
}), axis=1).fillna(0)

# Adding the event_date column back
expanded_data.insert(0, 'event_date', grouped_data['event_date'])

# Saving the new dataset to a CSV file
expanded_data.to_csv('/content/drive/MyDrive/Research/Bangladesh Movement Research/daily_event_summary.csv', index=False)

print("The new dataset has been created and saved as 'daily_event_summary.csv'.")
expanded_data
```

# **Division-wise Dataset Preparation**

```python
# Loading Raw Dataset
file_path = '/content/drive/MyDrive/Research/Bangladesh Movement Research/2024-06-05-2024-08-05-South_Asia-Bangladesh.csv'
data = pd.read_csv(file_path)

# Converting event_date to datetime
data['event_date'] = pd.to_datetime(data['event_date'])

# Creating complete date-division grid
all_dates = pd.date_range(start=data['event_date'].min(),
                         end=data['event_date'].max(),
                         freq='D')
all_divisions = data['admin1'].unique()
grid = pd.MultiIndex.from_product([all_dates, all_divisions],
                                 names=['event_date', 'admin1']).to_frame(index=False)

# Aggregating data by date and division
grouped = data.groupby(['event_date', 'admin1']).agg(
    total_fatalities=('fatalities', 'sum'),
    total_events=('event_id_cnty', 'count')
).reset_index()

# Mergig with complete grid (to include divisions with 0 events)
merged = pd.merge(grid, grouped, on=['event_date', 'admin1'], how='left').fillna(0)

# Adding population data (2023 estimates in millions)
population_mapping = {
    'Dhaka': 44.2,          # Millions
    'Chittagong': 33.2,
    'Rajshahi': 20.3,
    'Khulna': 17.8,
    'Barisal': 9.1,
    'Sylhet': 12.3,
    'Rangpur': 18.0,
    'Mymensingh': 12.3
}

merged['population'] = merged['admin1'].map(population_mapping)

# Converting population to absolute numbers (millions to actual count)
merged['population'] = merged['population'] * 1e6

# Ensuring proper column order ['admin1' is the Division]
final_columns = ['event_date', 'admin1', 'total_fatalities', 'total_events', 'population']
merged = merged[final_columns]

# Saving the final dataset
merged.to_csv('/content/drive/MyDrive/Research/Bangladesh Movement Research/daily_division_dataset.csv', index=False)

print("Dataset created with shape:", merged.shape)
merged.sample(5)
```
